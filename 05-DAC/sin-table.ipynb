```python
import numpy as np
U_sin = np.array([ 2.5, 3.3, 4.0, 4.5, 4.9, 5.0, 4.9, 4.5, 4.0, 3.3, 2.5, 1.7, 1.0, 0.5, 0.1, 0.0, 0.1, 0.5, 1.0, 1.7 ])
U_ref = 5
D = np.round(U_sin*15/5).astype(int)
```

```python

U_synth = np.zeros(len(D))
for i in range(len(D)):
	for j in range(4):
		U_synth[i] = U_synth[i]	+ ((D[i]>>(3-j))&0b1) / 2**(j+1)

for b, u in zip(D, 5*U_synth):
    print(f"{b} & 0x{b:01x} & {u:.2f} \\\\")
```
